```python
%%writefile app.py
import streamlit as st
import pdfplumber
import pandas as pd
import time
from keybert import KeyBERT

# Using centered layout for better aesthetics
st.set_page_config(page_title="KeyBERT-MiniLM-HighPrecision-Extractor", layout="centered")

# Custom CSS for UI centering and scrollable table
st.markdown("""
    <style>
    .main {
        max-width: 800px;
        margin: 0 auto;
    }
    .stTable {
        background-color: white;
        border-radius: 10px;
    }
    .table-container {
        max-height: 400px;
        overflow-y: auto;
        border: 1px solid #e6e9ef;
        border-radius: 10px;
        margin-bottom: 20px;
    }
    </style>
    """, unsafe_allow_html=True)

@st.cache_resource
def load_nlp_model():
    return KeyBERT(model='all-MiniLM-L6-v2')

kw_model = load_nlp_model()

def extract_data_from_multiple_pdfs(files):
    combined_text = ""
    for file in files:
        try:
            with pdfplumber.open(file) as pdf:
                for page in pdf.pages:
                    page_text = page.extract_text()
                    if page_text:
                        combined_text += page_text + " "
        except Exception as e:
            st.error(f"Error reading {file.name}: {e}")
    return combined_text

st.title("KeyBERT-MiniLM-HighPrecision-Extractor")
st.write("Upload one or more PDFs to extract significant keywords using Natural Language Processing.")

# Enabled multiple file selection
uploaded_files = st.file_uploader("Upload PDF Documents", type=['pdf'], accept_multiple_files=True)

if uploaded_files:
    # Process files automatically
    with st.status("Analyzing document content...", expanded=True) as status:
        document_text = extract_data_from_multiple_pdfs(uploaded_files)

        if document_text.strip():
            keywords = kw_model.extract_keywords(
                document_text,
                keyphrase_ngram_range=(1, 2),
                stop_words='english',
                use_mmr=True,
                diversity=0.6,
                top_n=30
            )
            status.update(label="Analysis Complete!", state="complete", expanded=False)

            # This part ensures results are shown without needing to click the status bar
            st.subheader(f"Extracted Results ({len(keywords)} Keywords)")

            df = pd.DataFrame(keywords, columns=['Keyword', 'Relevance Score'])
            df.index = df.index + 1

            st.markdown('<div class="table-container">', unsafe_allow_html=True)
            st.table(df.style.format({'Relevance Score': "{:.4f}"}))
            st.markdown('</div>', unsafe_allow_html=True)

            csv = df.to_csv(index=False).encode('utf-8')
            st.download_button(
                label="Download Keywords as CSV",
                data=csv,
                file_name=f'extracted_keywords.csv',
                mime='text/csv',
                use_container_width=True
            )
        else:
            st.error("Could not find any readable text in the uploaded PDFs.")
```
